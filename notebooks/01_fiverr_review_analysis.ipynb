````python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()
````

````python
def get_sentiment(text):
    score = analyzer.polarity_scores(str(text))["compound"]

    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"
````

````python
import re
import pandas as pd

def clean_review_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Replace line breaks/tabs with spaces
    text = re.sub(r"[\r\n\t]+", " ", text)

    # Remove excessive whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove leading/trailing spaces
    text = text.strip()

    return text


reviews["review_text_llm"] = reviews["review_text"].apply(clean_review_text)
````

````python
def clean_review_text(text):
    if pd.isna(text):
        return ""

    text = str(text)

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)

    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Replace line breaks/tabs with spaces
    text = re.sub(r"[\r\n\t]+", " ", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


reviews["review_text_llm"] = reviews["review_text"].apply(clean_review_text)
````

````python
from pydantic import BaseModel, Field
from typing import List, Literal


class ReviewAnalysis(BaseModel):
    sentiment: Literal[
        "Positive",
        "Neutral",
        "Negative",
        "Mixed"
    ]

    confidence: float = Field(
        ge=0,
        le=1
    )

    satisfaction: float = Field(
        ge=0,
        le=1
    )

    strengths: List[str]

    improvement_areas: List[str]

    topics: List[str]

    explanation: str

    recommendation: str
````

````python
SYSTEM_PROMPT = """
You are an expert AI analyst specializing in Fiverr client feedback and NLP.

Analyze the provided Fiverr client review.

Return ONLY valid JSON matching the requested schema.

Analyze:

1. sentiment
   - Positive
   - Neutral
   - Negative
   - Mixed

2. confidence
   Number between 0 and 1.

3. satisfaction
   Number between 0 and 1.

4. strengths
   Specific things the client liked or praised.

5. improvement_areas
   Specific problems, concerns, weaknesses, or areas that could be improved.

6. topics
   Select only relevant topics from:
   - Communication
   - Technical Quality
   - Delivery
   - Requirements
   - Reliability
   - Professionalism
   - Problem Solving
   - Revisions
   - Pricing
   - Other

7. explanation
   Brief explanation of the analysis.

8. recommendation
   Practical recommendation based only on the review.

IMPORTANT:
- Analyze only the review.
- Do not invent problems.
- Do not assume a problem exists.
- If no improvement area is mentioned or reasonably implied, return [].
- Do not use the Fiverr star rating to determine sentiment.
- Keep the analysis concise and evidence-based.
"""
````

````python
import json
import re

def clean_llm_json(text):
    """
    Clean JSON returned by an LLM.
    Handles ```json ... ``` and plain JSON.
    """

    text = text.strip()

    # Remove markdown code fence
    text = re.sub(r"^```json\s*", "", text, flags=re.IGNORECASE)
    text = re.sub(r"^```\s*", "", text)
    text = re.sub(r"\s*```$", "", text)

    return text.strip()
````

````python
def analyze_review(review_text):

    response = client.chat(
        model="gpt-oss:20b-cloud",
        messages=[
            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": review_text
            }
        ],
        format=ReviewAnalysis.model_json_schema()
    )

    raw_result = response["message"]["content"]

    cleaned_result = clean_llm_json(raw_result)

    analysis = ReviewAnalysis.model_validate_json(
        cleaned_result
    )

    return analysis
````

````python
import os
import json
import time
import pandas as pd
from datetime import datetime

OUTPUT_FILE = "outputs/llm_review_analysis.csv"

# Make sure output directory exists
os.makedirs("outputs", exist_ok=True)


def analyze_review(review_text):
    """
    Send one review to Ollama Cloud and return structured analysis.
    """

    response = client.chat(
        model="gpt-oss:20b-cloud",
        messages=[
            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": review_text
            }
        ],
        format=ReviewAnalysis.model_json_schema()
    )

    raw_result = response["message"]["content"]

    cleaned_result = clean_llm_json(raw_result)

    analysis = ReviewAnalysis.model_validate_json(
        cleaned_result
    )

    return analysis
````

````python
def save_analysis_result(review_id, analysis, status="success", error=None):

    result = {
        "review_id": review_id,
        "llm_sentiment": analysis.sentiment if analysis else None,
        "llm_confidence": analysis.confidence if analysis else None,
        "satisfaction": analysis.satisfaction if analysis else None,
        "strengths": json.dumps(
            analysis.strengths,
            ensure_ascii=False
        ) if analysis else None,
        "improvement_areas": json.dumps(
            analysis.improvement_areas,
            ensure_ascii=False
        ) if analysis else None,
        "topics": json.dumps(
            analysis.topics,
            ensure_ascii=False
        ) if analysis else None,
        "explanation": analysis.explanation if analysis else None,
        "recommendation": analysis.recommendation if analysis else None,
        "status": status,
        "error": error,
        "processed_at": datetime.now().isoformat()
    }

    result_df = pd.DataFrame([result])

    # Append to CSV
    if os.path.exists(OUTPUT_FILE):
        result_df.to_csv(
            OUTPUT_FILE,
            mode="a",
            header=False,
            index=False,
            encoding="utf-8-sig"
        )
    else:
        result_df.to_csv(
            OUTPUT_FILE,
            mode="w",
            header=True,
            index=False,
            encoding="utf-8-sig"
        )
````

````python
def save_analysis_result(review_id, analysis, status="success", error=None):

    result = {
        "review_id": review_id,
        "llm_sentiment": analysis.sentiment if analysis else None,
        "llm_confidence": analysis.confidence if analysis else None,
        "satisfaction": analysis.satisfaction if analysis else None,
        "strengths": json.dumps(
            analysis.strengths,
            ensure_ascii=False
        ) if analysis else None,
        "improvement_areas": json.dumps(
            analysis.improvement_areas,
            ensure_ascii=False
        ) if analysis else None,
        "topics": json.dumps(
            analysis.topics,
            ensure_ascii=False
        ) if analysis else None,
        "explanation": analysis.explanation if analysis else None,
        "recommendation": analysis.recommendation if analysis else None,
        "status": status,
        "error": error,
        "processed_at": datetime.now().isoformat()
    }

    result_df = pd.DataFrame([result])

    # Append to CSV
    if os.path.exists(OUTPUT_FILE):
        result_df.to_csv(
            OUTPUT_FILE,
            mode="a",
            header=False,
            index=False,
            encoding="utf-8-sig"
        )
    else:
        result_df.to_csv(
            OUTPUT_FILE,
            mode="w",
            header=True,
            index=False,
            encoding="utf-8-sig"
        )
````

````python
def process_all_reviews(reviews):

    # Check already processed reviews
    if os.path.exists(OUTPUT_FILE):

        existing = pd.read_csv(
            OUTPUT_FILE
        )

        processed_ids = set(
            existing.loc[
                existing["status"] == "success",
                "review_id"
            ]
        )

        print(
            f"Already processed: {len(processed_ids)} reviews"
        )

    else:

        processed_ids = set()

        print("Starting new analysis...")

    total = len(reviews)

    for index, row in reviews.iterrows():

        review_id = int(row["review_id"])

        # Skip already completed reviews
        if review_id in processed_ids:
            continue

        review_text = row["review_text_llm"]

        print(
            f"Processing {review_id}/{total}..."
        )

        try:

            # Skip empty reviews
            if not review_text or not review_text.strip():

                print(
                    f"Review {review_id}: empty text"
                )

                save_analysis_result(
                    review_id,
                    None,
                    status="skipped",
                    error="Empty review text"
                )

                continue

            # Analyze with Ollama
            analysis = analyze_review(
                review_text
            )

            # Save immediately
            save_analysis_result(
                review_id,
                analysis,
                status="success"
            )

            print(
                f"✓ Review {review_id}: "
                f"{analysis.sentiment}"
            )

        except Exception as e:

            error_message = str(e)

            print(
                f"✗ Review {review_id} failed: "
                f"{error_message}"
            )

            save_analysis_result(
                review_id,
                None,
                status="error",
                error=error_message
            )

        # Small delay between API calls
        time.sleep(0.5)

    print("\nAnalysis completed.")
````
